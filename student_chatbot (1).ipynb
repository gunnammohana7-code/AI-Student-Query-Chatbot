```python
# Complete improved chatbot

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
import gradio as gr

# 1. Training data
data = {
    "text": [
        "How can I apply for admission?",
        "What is the admission process?",
        "How do I get admission?",
        "What are the admission requirements?",
        "How can I join the college?",

        "What courses are available?",
        "What branches does the college offer?",
        "Which programs are offered?",
        "Tell me about the courses",
        "Which engineering branches are available?",

        "What is the college fee?",
        "How much is the BTech fee?",
        "Tell me about tuition fees",
        "How much do I need to pay?",
        "What are the fees for engineering?",

        "Are internships available?",
        "Does the college provide internships?",
        "How can I get an internship?",
        "Does the college offer internships?",
        "Tell me about internship opportunities",

        "What about placements?",
        "Does the college provide placements?",
        "Tell me about placement opportunities",
        "What are the placement opportunities?",
        "Does the college have campus placements?"
    ],

    "intent": [
        "admission", "admission", "admission", "admission", "admission",
        "courses", "courses", "courses", "courses", "courses",
        "fees", "fees", "fees", "fees", "fees",
        "internship", "internship", "internship", "internship", "internship",
        "placements", "placements", "placements", "placements", "placements"
    ]
}

df = pd.DataFrame(data)

# 2. Convert text into TF-IDF numbers
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["text"])

# 3. Train AI model
model = LogisticRegression()
model.fit(X, df["intent"])

# 4. Responses
responses = {
    "admission": "You can contact the college admission office for the admission process and requirements.",
    "courses": "The college offers different engineering courses and branches.",
    "fees": "The BTech fee depends on the course and admission category. Please contact the college office for the latest fee structure.",
    "internship": "Yes, students can explore internship opportunities through the college and external organizations.",
    "placements": "The college provides placement opportunities through its placement cell and recruiting companies."
}

# 5. Chatbot function
def chatbot(message):
    question_vector = vectorizer.transform([message])
    predicted_intent = model.predict(question_vector)[0]
    return responses[predicted_intent]

# 6. Create interface
demo = gr.Interface(
    fn=chatbot,
    inputs=gr.Textbox(
        label="Ask your question",
        placeholder="Example: What courses are available?"
    ),
    outputs=gr.Textbox(label="Chatbot"),
    title="🎓 AI Student Query Chatbot",
    description="Ask questions about admissions, courses, fees, internships and placements."
)

demo.launch(share=True)
```
